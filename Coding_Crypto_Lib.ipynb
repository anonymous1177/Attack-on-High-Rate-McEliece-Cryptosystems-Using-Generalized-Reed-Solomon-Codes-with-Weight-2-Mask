```python
def schur_product(v,w):
    if len(v) != len(w):
        raise Exception("Input Lengths have to be Equal")
    y=[]
    for i in range(len(v)):
        y.append(v[i]*w[i])
    y=vector(y)
    return y

def schur_triple_product(v, w, x):
    if len(v) != len(w) or len(w) != len(x):
        raise Exception("Input Lengths have to be Equal")
    y=[]
    for i in range(len(v)):
        y.append(v[i]*w[i]*x[i])
    y=vector(y)
    return y

def product_code(G_1, G_2):
    k_1=G_1.nrows()
    k_2=G_2.nrows()
    G_rows = []
    for i in range(k_1):
        for j in range(k_2):
            G_rows.append(schur_product(G_1[i], G_2[j]))
    G = matrix(G_rows)
    G_r = G.rref()
    return G_r[:G_r.rank()]

def square_code_generator(G):
    # this returns a matrix with k(k+1)/2 rows.
    k=G.nrows()
    H=[]
    for i in range(k):
        for j in range(i+1):
            H.append(schur_product(G[i], G[j]))
    return matrix(H)

def square_code_generator_full_rank(G):
    G_square = square_code_generator(G)
    G_s_r = G_square.rref()
    return G_s_r[:G_s_r.rank()]


def random_generator_matrix(F, k, n):
    mat_space = MatrixSpace(F, k, n)
    G = mat_space.random_element()
    while G.rank() != k:
        G = mat_space.random_element()
    return G

def random_permutation_matrix(n):
    return Permutations(n).random_element().to_matrix()

def random_diagonal_entries_and_matrix(F, n):
    # has non-zero diagonal entries
    inv_elts = F.list()[1:]
    diag_elts = []
    for i in range(n):
        rand = randint(0, q-2)
        diag_elts.append(inv_elts[rand])
    diag_elts
    D = diagonal_matrix(diag_elts)
    return diag_elts, D


def cube_code(G):
    k=G.nrows()
    cube_list =[]
    for i in range(k):
        for j in range(i+1):
            for k in range(j+1):
                cube_list.append(schur_triple_product(G[i], G[j], G[k]))
    cube_unreduced = matrix(cube_list)
    cube_reduced = cube_unreduced.rref()
    return cube_reduced[:cube_reduced.rank()]
```
